**DataTypes**

* Datatypes are a classification of data which tell the compiler how the programmer intends to use the data.
* They determines the type of operations that can be performed on the data, the values that data can take, and the amount of memory needed to store the data.


**Importance of DataType**

* DataTypes ensure that data is stored in an efficient way.
* They help in performing correct operations on data.
* Proper use of data types can prevent errors and bugs in the program.

```python
age=21
type(age)
```

```python
## Floating point datatypes
height=5.6
print(height)
print(type(height))
```

```python
## String datatypes
name="Adarsh"
print(name)
print(type(name))
```

```python
## Boolean
is_true=True
type(is_true)
```

```python
a=10
b=10
type(a==b)
```

```python
## Common errors
result="Hello"+5
```

```python
result="Hello"+str(5)
print(result)
```

```python
## Type casting
a="31.2"
b=float(a)
print(b)
print(type(b))
```

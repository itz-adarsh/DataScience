**Variables**

Variables are fundamental elements in programming used to store data can be referenced and manipulated in a program. In python, variables are created when you assign a value to them.

```python
## Declaring and Assigning varibales

age=21
height=5.6
name="Adarsh"
is_student=True

print("Age:",age)
print("Height:",height)
print("Name:",name)
print("Is Student:",is_student)
```

```python
## Understanding variables types
## Python is dynamically typed , type of variables is determind at runtime

age=21 # int
height=5.6 # float
name="Adarsh" # string

print(type(name))
```

```python
## type checking and conversion
type(height)
```

```python
## Dynamic Typing
## Python allows the type of a varible to change as the program executes
var=10 #int
print(var,type(var))

var="Hello"
print(var,type(var))

var=3.14
print(var,type(var))
```

```python
## Simple calculator

num1=float(input("Enter first number: "))
num2=float(input("Enter second number: "))

sum=num1+num2
difference=num1-num2
product=num1*num2
quotient=num1/num2

print("Sum:",sum)
print("Differnce:",difference)
print("Product:",product)
print("Quotient:",quotient)
```

***Lambda Functions in Python***

Lambda function are small anonymous function defined using the ***lambda*** keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguemnts to higher-order functions

```python
## Syntax 
lambda arguments: expression
```

```python
def addition(x, y):
    return x + y
addition(2,4)
```

```python
addition=lambda a,b:a+b ## here we are using lambda function to perform addition and 
print(type(addition))
print(addition(2,4))
```

```python
def even(num):
    if num%2==0:
        return True
    else:
        return False
print(even(4))
```

```python
even=lambda num:num%2==0 ## here we are using lambda function to check if the number is even or not
print(even(4))
```

```python
def addition(x,y,z):
    return x+y+z
addition(2,4,6)
```

```python
sum = lambda a,b,c:a+b+c ## here we are using lambda function to perform addition of three numbers
print(sum(2,4,6))
```

```python
numbers=[1,2,3,4,5,6]
def sqaure(number):
    return number**2
sqaure(2)
```

```python
## map()-applies a function to items in a list
list(map(lambda x:x**2,numbers))
```

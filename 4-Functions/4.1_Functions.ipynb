```python
# Why functions?
num=24
if num%2==0:
    print("Even")
else:
    print("Odd")
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if num%2==0:
        print("The number is even")
    else:
        print("The number is odd")
```

```python
## call this function
even_or_odd(24)
```

```python
## Function with multiple parameters

def add(a,b):
    return a+b
result=add(2,4)
print(result)
```

```python
## Default parameters

def greet(name):
    print(f"Hello {name} welcome to the paradise")
    
greet("Adarsh")
```

```python
def greet(name="Adarsh"):  ## here we have set a default value for the parameter name 
    print(f"Hello {name} welcome to the paradise")
    
greet() ## here we have not passed any value to the parameter name so it will take the default value "Adarsh" and print "Hello Adarsh welcome to the paradise"
greet("John") ## here we have overridden the default value of the parameter name by passing a new value "John" to it. so it will print "Hello John welcome to the paradise" instead of "Hello Adarsh welcome to the paradise"
```

```python
## Variable length arguments
## Positional arguments - It means that we can pass any number of arguments to a function and it will accept them as a tuple. We use *args to accept variable length positional arguments.
def print_numbers(*adarsh):  ## here we have used *args to accept variable length arguments. args is a convention but you can use any name instead of args.
    for numbers in adarsh:
        print(numbers)
print_numbers(1,2,3,4,5,"Adarsh")
```

```python
## best practice is 
def print_numbers(*args): ## here we have used *args to accept variable length arguments. args is a convention but you can use any name instead of args.
    for numbers in args:
        print(numbers)
print_numbers(1,2,3,4,5,"Adarsh")
```

```python
## Keyword arguments - It means that we can specify the name of the parameter while passing the argument to the function. This is useful when we have multiple parameters in a function and we want to pass the arguments in any order. In keyword agruments all the value should be in key value pairs
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
print_details(name="Adarsh",age=22,city="Gaya")
```

```python
## combination of positional and keyword arguments
def print_info(*args,**kwargs):
    for numbers in args:
        print(f"Positional argument: {numbers}")
    for key,value in kwargs.items():
        print(f"Keyword argument: {key} = {value}")
print_info(1,2,3,name="Adarsh",age=22,city="Gaya")
```

```python
## Return statement
def multiply(a,b):
    return a*b
result=multiply(2,4)
print(result)
```

```python
## we can also return multiple values from a function by separating them with a comma. The return statement will return a tuple containing all the values.
def multiply(a,b):
    return a*b,a
result=multiply(2,4)
print(result)
```

```python
## Program to find the greatest of three number
def greatest(a,b,c):
    if a>b and a>c:
        return a
    elif b>a and b>c:
        return b
    else:
        return c
print(greatest(1,2,3))
```

```python
# sum of first n natural number using recursion
def sum(n):
    if(n==1):
        return 1
    else:
        return sum(n-1)+n
print(sum(4))
```

```python
def pattern(n):
    if n==0:
        return
    print("*"*n)
    pattern(n-1)
pattern(5)
```

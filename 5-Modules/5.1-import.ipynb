***Importing Modules in Pyhton : Modules and Packages***

In Python, modules and packages help organizes and reuse code. Here's a comprehensive guide on how to import them

```python
import math
print(math.sqrt(16))
print(math.pi)
```

```python
from math import sqrt,pi
print(sqrt(16))
print(sqrt(25))
print(pi)
```

```python
from math import *  # to import all from math
print(sqrt(16))
print(pi)
```

```python
## creating own packages
from package.maths import add
add(2,3)
```

```python
from package import maths
maths.add(2,3)
```

```python
import re
text="My numner is 78906445675"
result=re.findall(r'\d+',text)   ## \d+ means one or more digits i.e it will find all the numbers in the text and return them as a list
print(result)            ## also on d+ we can use \d{10} to find only 10 digit numbers in the text . Here r is used to indicate that the string is a raw string and it will not treat backslashes as escape characters.
```

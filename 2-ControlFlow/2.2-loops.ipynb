```python
## while loop - continues to execute as long as the condition is true
count=0
while count<5:
    print(count)
    count=count+1
```

```python
count = 0 
while(count%2==0):
    print(count)
    count = count+1
```

```python
## loop control statement

## Break - The break statement exits the loop prematurely
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## Continue - Continue statement skips the current iteration and continues with the next
for i in range(10):
    if(i%2==0):
        continue
    print(i)
```

```python
## pass - the pass statement is null operation it does nothing
for i in range(5):
    if i ==3:
        pass
    print(i)
```

```python
## Nested loop - loop inside a loop
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
## 1. Calculate the sum of first N natural number using a while loop and for loop
## while loop
n=10
sum=0
count=1
while(count<=n):
    sum = sum + count
    count = count + 1
print("Sum of first 10 natural number:",sum)
```

```python
## for loop
n=10
sum=0
for i in range(11):
    sum=sum+i
print(sum)
```

```python
## 2. Display the prime number between 1 and 100
for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)
```

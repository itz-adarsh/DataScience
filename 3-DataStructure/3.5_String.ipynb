**Strings**
* String is a datatype in python.
* String is a sequence of character enclosed in quotes.
* String is immutable means we cannot change any string after making it 



```python
a = "Adarsh"
b = 'Adarsh'
c = '''Adarsh'''
print(a[2])
print(a[-1])
```

```python
### String slicing in Python
name = "Adarsh"
nameshort = name[0:3]  ## Slicing the string to get the first three characters. Start from index 0 all the way to index 3 (not inclusive).
print(nameshort)  # Output: Ada

charcter1 =  name[1]
print(charcter1)  # Output: d

print(name[-4:-1]) ## Negative indexing in strings

print(name[:2])  ## Omitting the starting index it is same as print(name[0:2])
print(name[2:])  ## Omitting the ending index it is same as print(name[2:len(name)])

print(name[:])  ## Omitting both the starting and ending index it is same as print(name[0:len(name)])

print(name[1:6:2])  ## Slicing with step value. Here step value is 2
print(name[::2])  ## Omitting the starting and ending index with step value. Here step value is 2
```

**String Functions** 

```python
name = "Adarsh Kumar"
print(len(name))  ## Length of the string
print(name.endswith("ar"))  ## Check if the string ends with a particular substring
print(name.startswith("Ad"))  ## Check if the string starts with a particular substring
print(name.capitalize())  ## Capitalize the first character of the string
print(name.lower())  ## Convert the string to lowercase
print(name.upper())  ## Convert the string to uppercase
print(name.find("sh"))  ## Find the index of a particular substring
```

```python
name = "Adarsh Kumar"
replced_name = name.replace("Kumar", "Verma")  ## Replace a substring with another substring
print(replced_name)  # Output: Adarsh Verma
```

```python
## Escape sequneces in strings
print("Hello \n World")  ## New line
print("Hello \t World")  ## Tab space
print("Hello \\ World")  ## Backslash
print("Hello \' World")  ## Single quote
print("Hello \" World")  ## Double quote
print("Hello \r World")  ## Carriage return
print("Hello \b World")  ## Backspace   
print("Hello \f World")  ## Form feed
print("Hello \v World")  ## Vertical tab
```

```python
print("Hii"*(3//2))  ## Repeating the string 3 times and then dividing it by 2 to get the first half of the repeated string
```

```python
## String formatting 
name = "Adarsh"
age = 21

print("Name: %s Age: %d" % (name, age)) ## old style
print("Name: {} Age: {}".format(name, age)) ## fromat method 
print(f"Name: {name} Age: {age}") ## Best method using f-strings
```

***Introduction to Tuples***
* Tuples are ordered collections of items that are immutable. They are similar to lists, but their immutability makes them different.

```python
## Creating a tuple 
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst = list() ## here we are creating an empty list using the list() constructor
print(type(lst))
tpl = tuple()  ## here we are creating an empty tuple using the tuple() constructor
print(type(tpl))
```

```python
numbers=tuple([1,2,3,4,5,6])  ## here we are creating a tuple from a list using the tuple() constructor
print(numbers)
print(type(numbers))
```

```python
number = list((1,2,3,4,5,6))
print(number)
print(type(number))
```

```python
mixed_tuples = (1,"Hello World",3.14,True)
print(mixed_tuples)
```

```python
## Tuples operations 
numbers = (1,2,3,4,5,6)
mixed_tuples = (1,"Hello World",3.14,True)
```

```python
## Concatenation of tuples
Concat_Tuples = numbers + mixed_tuples
print(Concat_Tuples)
```

```python
## Immutable nature of tuple
lst = [1,2,3,4,5]
print(lst)

lst[1]="Krish"
print(lst)
```

```python
numbers=(1,2,3,4,5,6)
numbers[1]="Krish"  ## It shows tuples are immutable 
```

```python
numbers
```

```python
## Tuple methods
print(numbers.count(1))
print(numbers.index(3)) ## three is present at which index
```

```python
## Packing and unpacking tuple
packed_tuples = 1,"Hello",3.14  ## here we are packing the values into a tuple without using parenthesis by default it is considered as a tuple
print(packed_tuples)    
```

```python
## Unpacking the tuple
a,b,c = packed_tuples ## Unpacking means we are assigning the values of the tuple to the variables a,b,c
print(a)
print(b)         ## Unpacking is possible only when the number of variables is equal to the number of values in the tuple
print(c)
```

```python
## Unpacking with * operator
numbers = (1,2,3,4,5,6)
a,*b,c = numbers  ## here we are unpacking the first and last value of the tuple to the variables a and c respectively and the rest of the values to the variable b using the * operator
print(a)
print(b)
print(c)
```

```python
## Nested List
lst = [[1,2,3,4],[6,7,8,9],[1,"Hello",3.14,True]]
print(lst[0]) ## here we are accessing the first list from the nested list
print(lst[0][2]) ## here we are accessing the third element from the first list of the nested list
print(lst[0][0:3]) ## here we are slicing the first three elements from the first list of the nested list
```

```python
lst = [[1,2,3,4],[6,7,8,9],(1,"Hello",3.14,True)] ## here we are creating a nested list which contains a tuple as its third element
print(lst[2][0:3]) ## here we are slicing the first three elements from the tuple which is the third element of the nested list
```

```python
## Nested Tuples
nested_tuple = ((1,2,3,4),(6,7,8,9),(1,"Hello",3.14,True)) ## here we are creating a nested tuple which contains three tuples as its elements
print(nested_tuple)
print(nested_tuple[0]) ## here we are accessing the first tuple from the nested tuple
print(nested_tuple[0][2]) ## here we are accessing the third element from the first tuple of the nested tuple
print(nested_tuple[0][0:3]) ## here
```

```python
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ") ## here we are printing all the elements of the nested tuple in a single line using nested for loop and end parameter of print function
```

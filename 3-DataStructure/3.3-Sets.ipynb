##### Sets
Sets are built-in data type in Python used to store collections of unique items. They are unordered, meaning that the elements do not follow specific order, and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations like union, intersection, difference and symmetric difference.

```python
## create a set
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set()
print(type(my_empty_set))
```

```python
my_set=set([1,2,3,4,5,6])
print(my_set)
```

```python
my_set1={1,2,3,4,5,5,6,7,4}  ## here it is not printing duplicate element
print(my_set)
```

```python
## Basic set operation 
## Adding and removing element
my_set.add(7)
print(my_set)
my_set.add(7)
print(my_set)
```

```python
## Remove the elements from a set
my_set.remove(3)
print(my_set)
```

```python
my_set.remove(10)
print(my_set)
```

```python
my_set.discard(11)  ## discard will perform operation whether it is present or not but remove works only when it is present
print(my_set)
```

```python
## pop
removed_elemnet=my_set.pop()
print(removed_elemnet)
print(my_set)
```

```python
## clear all the elements 
my_set.clear()
print(my_set)
```

```python
## Set membership test
my_set={1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
## Mathematical operations

set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## union 
union_set = set1.union(set2)
print("Union is:",union_set)

## intersection

intersection_set = set1.intersection(set2)
print("Intersection is:",intersection_set)

## Intersection update means it will perfrom the operation and update the given set
set1.intersection_update(set2)
print(set1)
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## difference
set_diff=set1.difference(set2)
print("Difference is:", set_diff)
```

```python
set2.difference(set1)
```

```python
## Symmetric difference
syy_diff_set=set1.symmetric_difference(set2)
print("Symmetric difference is:",syy_diff_set)
```

```python
## Set methods
set1={1,2,3}
set2={3,4,5}

## is subset
print(set1.issubset(set2))

## is superset
print(set1.issuperset(set2))
```

```python
lst = [1,2,2,3,4,4,5]
set(lst)
```

```python
## Counting unique words in text

text="In this tutorial we are discussing about sets"
words=text.split()

## convert list of words to set to get unique words

unique_words=set(words)
print(unique_words)
print(len(unique_words))
```

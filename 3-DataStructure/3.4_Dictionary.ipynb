##### Introduction to Dictionaries

Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable (e.g. strings, numbers, tuples) while values can be of any type.

```python
## Create Dictionaries
empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student = {"name":"Adarsh","age":22,"grade":10}
print(student)
```

```python
student = {"name":"Adarsh","age":22,"name":10}  ## here the key "name" is repeated, so the value will be updated to 10
print(student)
```

```python
student = {"name":"Adarsh","age":22,"grade":'A'}
print(student)
```

```python
## Accessing dictionary elements
print(student['grade'])
print(student['age'])

## ## Accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name',"Not Available")) ## here we have given default value not avialable if element is not present in dictionary
```

```python
## Modifying Dictionary Elements
## Dictionary are mutable, so you can add, update or delete elements
print(student)
```

```python
student["age"]=25  ## updated the value for key
print(student)
student["address"]="India" ## added a new key and value 
print(student)
```

```python
del student['grade'] ## delete key and value pair
print(student)
```

```python
## Dictionary method 

keys = student.keys() ## get all the keys
print(keys)

values=student.values() ## get all values
print(values)

items = student.items()  ## get all key value pairs
print(items)
```

```python
## Shawllow copy -- when we create a new variable and assign it to an existing dictionary, it creates a reference to the same dictionary in memory. So any changes made to the dictionary through either variable will affect both variables because they are pointing to the same dictionary.\
student_copy = student
print(student)
print(student_copy)
```

```python
student["name"]="Adarsh1"  ## here we have updated the value for key "name" in student dictionary, so it will also update in student_copy because both are pointing to same dictionary in memory so we need shallow copy to create a new dictionary with same key value pairs but it will not affect the original dictionary when we make changes to the new dictionary.
print(student)
print(student_copy)
```

```python
student_copy1=student.copy() ## here we have created a shallow copy of student dictionary, so any changes made to student_copy1 will not affect the original student dictionary because it is a new dictionary with same key value pairs but it is not pointing to the same dictionary in memory.
print(student_copy1)
print(student)
```

```python
student["name"]="Adarsh2" ## here we have updated the value for key "name" in student dictionary, so it will not update in student_copy1 because it is a new dictionary with same key value pairs but it is not pointing to the same dictionary in memory.
print(student)
print(student_copy1)
```

```python
## Iterating over dictionary
## you can use loops to iterate over the keys, values or key value pairs in a dictionary

## Iterating over keys
for keys in student.keys():
    print(keys)
```

```python
## iterate over values
for values in student.values():
    print(values)
```

```python
## iterating over key value pairs
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
## Nested Dictionary

students={
    "student1":{"name":"Adarsh","age":22},
    "student2":{"name":"Peter","age":35}
}
print(students)
```

```python
## Accessing nested dictionary elements
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
## Iterating over nested dictionary
for student_id,student_info in students.items():
    print(f"{student_id},{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary comprehension
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## Conditional dictionary comprehension
evens = {x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Practical examples
## use a dictionary to count the frequency of elements in list

numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}
for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## Merge two dictionary into one

dict1={"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```
